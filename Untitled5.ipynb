## Organizar Arquivos de Imagem por Data

Este notebook irá ajudá-lo a organizar seus arquivos de imagem baixados, movendo-os para subpastas estruturadas por ano, mês e dia.

```python
import os
import shutil
from datetime import datetime

# --- Configuração ---
# Especifique o diretório onde suas imagens estão atualmente.
# IMPORTANTE: Se suas imagens estão em um arquivo .zip, você precisará descompactá-lo primeiro.
# Por exemplo, se '/content/PIX_WhatsApp 2026-09-08 at 17.33.09.zip' contém as imagens,
# descompacte-o em um novo diretório, como '/content/imagens_descompactadas'.
# Altere 'caminho/para/suas/imagens_baixadas' para o diretório real contendo as imagens.
source_directory = '/content/imagens_descompactadas' # Exemplo: Crie este diretório e descompacte o .zip aqui

# Especifique o diretório onde você quer que as imagens organizadas sejam salvas.
# Se '/content/Resultado' é um arquivo, você precisa removê-lo antes de criar um diretório com o mesmo nome.
destination_directory = '/content/Resultado_Organizado' # Mude para um novo diretório, ou garanta que '/content/Resultado' seja um diretório vazio.

# Verifique se o destination_directory existe e é um arquivo; se sim, avise ou remova.
if os.path.exists(destination_directory) and not os.path.isdir(destination_directory):
    print(f"Atenção: Um arquivo com o nome '{destination_directory}' já existe. Por favor, remova-o manualmente ou escolha um nome de diretório diferente.")
else:
    # Crie o diretório de destino se ele não existir
    os.makedirs(destination_directory, exist_ok=True)
    print(f"Diretório de origem: {source_directory}")
    print(f"Diretório de destino: {destination_directory}")
```

### Função para Obter a Data do Arquivo

Tentaremos obter a data de modificação do arquivo como a data de referência para organização. Isso é geralmente confiável para arquivos baixados.

```python
def get_file_date(filepath):
    """
    Tenta obter a data de modificação de um arquivo.
    """
    try:
        # Obter o timestamp da última modificação do arquivo
        timestamp = os.path.getmtime(filepath)
        return datetime.fromtimestamp(timestamp)
    except OSError:
        # Em caso de erro (ex: arquivo não encontrado), retorna None
        return None

# Exemplo de uso da função (se você quiser testar)
# file_date = get_file_date(os.path.join(source_directory, 'sua_imagem.jpg'))
# if file_date:
#     print(f"Data do arquivo: {file_date.strftime('%Y-%m-%d')}")
```

### Organizando os Arquivos

Este código irá iterar sobre todos os arquivos no diretório de origem, determinar sua data e movê-los para a estrutura de diretórios `ANO/MÊS/DIA` no diretório de destino. Apenas os arquivos de imagem comuns (jpg, png, gif, bmp, jpeg, tiff) serão processados.

```python
supported_image_extensions = ('.jpg', '.png', '.gif', '.bmp', '.jpeg', '.tiff')

# Iterar sobre todos os arquivos no diretório de origem
for filename in os.listdir(source_directory):
    filepath = os.path.join(source_directory, filename)

    # Verificar se é um arquivo e se é um tipo de imagem suportado
    if os.path.isfile(filepath) and filename.lower().endswith(supported_image_extensions):
        file_date = get_file_date(filepath)

        if file_date:
            year = file_date.year
            month = file_date.month
            day = file_date.day

            # Criar o caminho do diretório de destino baseado na data
            target_subdir = os.path.join(destination_directory, str(year), f'{month:02d}', f'{day:02d}')
            os.makedirs(target_subdir, exist_ok=True)

            # Definir o caminho completo do arquivo de destino
            destination_filepath = os.path.join(target_subdir, filename)

            # Mover o arquivo
            try:
                shutil.move(filepath, destination_filepath)
                print(f"Movido: {filename} -> {target_subdir}/")
            except shutil.Error as e:
                print(f"Erro ao mover {filename}: {e}")
            except Exception as e:
                print(f"Erro inesperado com {filename}: {e}")
        else:
            print(f"Não foi possível obter a data para: {filename}. Pulando.")
    elif os.path.isfile(filepath):
        print(f"Ignorando arquivo não-imagem: {filename}")

print("Organização de arquivos concluída!")
```

## Montando o Banco de Dados de Imagens (DataFrame)

Agora que as imagens estão organizadas, vamos criar um DataFrame pandas que servirá como nosso 'banco de dados'. Ele conterá o caminho completo, o nome do arquivo e a data de cada imagem, permitindo consultas fáceis.

```python
import pandas as pd
import os
from datetime import datetime

image_data = []

# Iterar sobre o diretório de destino (onde as imagens foram organizadas)
for root, _, files in os.walk(destination_directory):
    for file in files:
        if file.lower().endswith(supported_image_extensions):
            filepath = os.path.join(root, file)

            # Tentar extrair a data do caminho do diretório (ANO/MÊS/DIA)
            try:
                path_parts = root.split(os.sep)
                day = int(path_parts[-1])
                month = int(path_parts[-2])
                year = int(path_parts[-3])
                file_date = datetime(year, month, day).date()
            except (ValueError, IndexError):
                # Se a data não puder ser extraída do caminho, use a data de modificação
                file_date = datetime.fromtimestamp(os.path.getmtime(filepath)).date()

            image_data.append({
                'filepath': filepath,
                'filename': file,
                'date': file_date
            })

image_df = pd.DataFrame(image_data)

# Adiciona uma verificação para garantir que o DataFrame não esteja vazio antes de tentar converter a coluna 'date'
if not image_df.empty:
    image_df['date'] = pd.to_datetime(image_df['date']) # Garante que a coluna de data seja datetime
    print(f"Banco de dados de imagens criado com {len(image_df)} entradas.")
    display(image_df.head())
else:
    print("Nenhuma imagem encontrada para montar o banco de dados. Verifique se o diretório de destino contém imagens organizadas.")
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output, Image, HTML
from google.colab import files # Importando para a funcionalidade de download
import matplotlib.pyplot as plt # Importando para gráficos
import seaborn as sns # Importando para gráficos

# Função para buscar imagens por nome de arquivo
def search_images_by_filename(search_string):
    if search_string:
        return image_df[image_df['filename'].str.contains(search_string, case=False, na=False)]
    return image_df

# Função para buscar imagens por intervalo de datas
def search_images_by_date_range(start_date, end_date):
    if start_date and end_date:
        return image_df[(image_df['date'] >= pd.to_datetime(start_date)) & (image_df['date'] <= pd.to_datetime(end_date))]
    return image_df

# Widgets de entrada
filename_input = widgets.Text(
    value='',
    placeholder='Digite parte do nome do arquivo',
    description='Nome do Arquivo:',
    disabled=False
)

start_date_picker = widgets.DatePicker(
    description='Data Inicial:',
    disabled=False
)

end_date_picker = widgets.DatePicker(
    description='Data Final:',
    disabled=False
)

search_button = widgets.Button(description='Buscar Imagens')
clear_button = widgets.Button(description='Limpar Consulta') # Novo botão de limpar

# Área de saída principal para os resultados da busca e o gráfico
output_area = widgets.Output()
# Nova área de saída dedicada para a visualização de imagens
image_preview_output = widgets.Output()

def on_search_button_clicked(b):
    with output_area:
        clear_output()
        # Limpa também a área de pré-visualização de imagem antes de uma nova busca
        with image_preview_output:
            clear_output()

        filtered_df = image_df

        # Filtra por nome de arquivo
        if filename_input.value:
            filtered_df = search_images_by_filename(filename_input.value)

        # Filtra por intervalo de datas (se houver datas selecionadas e válidas)
        if start_date_picker.value and end_date_picker.value:
            # Certifique-se de que o filtro de data seja aplicado ao DataFrame já filtrado por nome de arquivo
            filtered_df = filtered_df[(filtered_df['date'] >= pd.to_datetime(start_date_picker.value)) & (filtered_df['date'] <= pd.to_datetime(end_date_picker.value))]

        if not filtered_df.empty:
            display(f"Encontradas {len(filtered_df)} imagens:")

            # Exibir informações do arquivo e botões para cada imagem
            for index, row in filtered_df.iterrows():
                # Informações do arquivo
                file_info_text = f"<b>Nome:</b> {row['filename']}<br><b>Data:</b> {row['date'].strftime('%Y-%m-%d')}"
                file_info_widget = widgets.HTML(value=file_info_text, layout=widgets.Layout(width='60%', padding='0 10px')) # Adiciona padding para espaçamento

                hbox_buttons = widgets.HBox([]) # Para agrupar botões de cada imagem

                # Cria um botão de download para a imagem
                download_button = widgets.Button(description="Baixar") # Descrição mais curta

                # Define a função de clique para este botão específico
                def download_image_clicked(btn, filepath=row['filepath'], filename=row['filename']):
                    # Não usa clear_output aqui para não apagar a lista de resultados
                    print(f"Preparando para baixar: {filename}...") # Mensagem aparece no console Colab
                    try:
                        files.download(filepath)
                        print(f"{filename} baixado com sucesso.")
                    except Exception as e:
                        print(f"Erro ao baixar {filename}: {e}")

                download_button.on_click(download_image_clicked)
                hbox_buttons.children += (download_button,)

                # Novo: Botão para visualizar a imagem
                view_button = widgets.Button(description="Visualizar") # Descrição mais curta

                def view_image_clicked(btn, filepath=row['filepath'], filename=row['filename']):
                    with image_preview_output: # Usa a área dedicada para a pré-visualização
                        clear_output(wait=True) # Limpa apenas a área de pré-visualização
                        display(f"Visualizando: {filename}")
                        try:
                            display(Image(filename=filepath, width=400)) # Ajuste a largura conforme necessário
                        except Exception as e:
                            print(f"Erro ao exibir a imagem {filename}: {e}")

                view_button.on_click(view_image_clicked)
                hbox_buttons.children += (view_button,)

                # Combina as informações do arquivo e os botões em um HBox para cada item
                # Adiciona borda e padding para melhor visualização de cada 'linha' de resultado
                item_display = widgets.HBox([
                    file_info_widget,
                    hbox_buttons
                ], layout=widgets.Layout(border='1px solid lightgray', padding='5px', margin='2px 0'))
                display(item_display) # Exibe a linha completa (info + botões)

            # Adicionar visualização de imagens por data
            daily_counts = filtered_df.groupby(filtered_df['date'].dt.date).size().reset_index(name='count')
            daily_counts.columns = ['date', 'count']

            fig, ax = plt.subplots(figsize=(10, 5))
            sns.barplot(x='date', y='count', data=daily_counts, ax=ax, palette='viridis')
            ax.set_title('Número de Imagens por Data (Resultados Filtrados)')
            ax.set_xlabel('Data')
            ax.set_ylabel('Número de Imagens')
            ax.tick_params(axis='x', rotation=45)
            plt.tight_layout()
            plt.show()

        else:
            print("Nenhuma imagem encontrada com os critérios especificados.")

def on_clear_button_clicked(b):
    with output_area:
        clear_output()
        with image_preview_output:
            clear_output()
        filename_input.value = ''
        start_date_picker.value = None
        end_date_picker.value = None

search_button.on_click(on_search_button_clicked)
clear_button.on_click(on_clear_button_clicked) # Conecta o botão de limpar à função

# Organiza os widgets em um layout
display(
    widgets.VBox([
        widgets.HBox([filename_input]),
        widgets.HBox([start_date_picker, end_date_picker]),
        widgets.HBox([search_button, clear_button]), # Adiciona o novo botão ao layout
        image_preview_output, # Exibe a área de pré-visualização de imagem
        output_area # Exibe a área de resultados da busca
    ])
)
```
